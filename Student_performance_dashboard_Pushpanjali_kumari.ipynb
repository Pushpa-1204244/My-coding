```python
import pandas as pd
df = pd.read_excel('/content/student_marks.CSV file.xlsx')
display(df)
```

Total & percenteage


```python
subject_columns = ['Math', 'Science', 'English', 'Social Studies', 'Computer Science']
df['Total'] = df[subject_columns].sum(axis=1)
max_total_score = len(subject_columns) * 100
df['Percentage'] = (df['Total'] / max_total_score) * 100
display(df)
```

Grades

```python
if 'Total' not in df.columns or 'Percentage' not in df.columns:
    subject_columns = ['Math', 'Science', 'English', 'Social Studies', 'Computer Science']
    df['Total'] = df[subject_columns].sum(axis=1)
    max_total_score = len(subject_columns) * 100
    df['Percentage'] = (df['Total'] / max_total_score) * 100

def calculate_grade(percentage):
  if percentage>=90:
    return 'A'
  elif percentage>=80:
    return 'B'
  elif percentage>=70:
    return 'C'
  elif percentage>=60:
    return 'D'
  else:
    return 'F'
df['Grade']=df['Percentage'].apply(calculate_grade)
display(df)
```

Top 5 and bottom 5 students

```python

top_5_students = df.sort_values(by='Total', ascending=False).head(5)
print("Top 5 Students:")
display(top_5_students)

bottom_5_students = df.sort_values(by='Total', ascending=True).head(5)
print("\nBottom 5 Students:")
display(bottom_5_students)
```

Subject wise class average

```python
subject_columns = ['Math', 'Science', 'English', 'Social Studies', 'Computer Science']
class_averages = df[subject_columns].mean()
print("Subject-wise Class Averages:")
print("="*60)
display(class_averages)
```

To count no. of students failed and passed

```python
passing_marks = 40

print("Subject-wise Pass and Fail Count:")
print("="*60)

for subject in subject_columns:
    passed_students = df[df[subject] >= passing_marks].shape[0]
    failed_students = df[df[subject] < passing_marks].shape[0]
    print(f"Subject: {subject}")
    print(f"  Passed: {passed_students} students")
    print(f"  Failed: {failed_students} students")
    print("-"*30)
```

Histogram showing the student percentage

```python
import pandas as pd
import matplotlib.pyplot as plt

df_percentage=pd.DataFrame(df['Percentage'])
plt.figure(figsize=(10,6))
plt.hist(df_percentage["Percentage"],bins=5,edgecolor="black",color='steelblue')
plt.title("Student percentage Distribution")
plt.xlabel("Percentage")
plt.ylabel("No. of students")
plt.tight_layout()
plt.show()
```

Bar chart to claculate subject wise average

```python
import pandas as pd
import matplotlib.pyplot as plt

subject_columns = ['Math', 'Science', 'English', 'Social Studies', 'Computer Science']
class_averages = df[subject_columns].mean()

plt.figure(figsize=(10,6))
plt.bar(x=class_averages.index, height=class_averages.values, edgecolor="black", color='blue')
plt.title("Subject-wise Class Averages")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Subject with Highest Failure Rate

```python
passing_marks = 40
highest_failure_count = -1
subject_with_highest_failure = ''

print("Subject with highest failure rate")
print("="*60)

for subject in subject_columns:
    failed_students_count = df[df[subject] < passing_marks].shape[0]
    print(f"Subject: {subject}, Failed Students: {failed_students_count}")

    if failed_students_count > highest_failure_count:
        highest_failure_count = failed_students_count
        subject_with_highest_failure = subject

print("="*60)
print(f"The subject with the highest failure rate is: {subject_with_highest_failure} with {highest_failure_count} failed students.")
```

```python
passing_marks = 40
df['Subjects_Passed_Count'] = df[subject_columns].apply(lambda x: (x >= passing_marks).sum(), axis=1)
df['Subject_Pass_Percentage'] = (df['Subjects_Passed_Count'] / len(subject_columns)) * 100
summary_report = df[['Student Name', 'Total', 'Percentage', 'Grade', 'Subject_Pass_Percentage']]
display(summary_report)
```

**Total Pass and Fail counts** **in all subjects**

```python

total_passed_students = 0
total_failed_students = 0

for subject in subject_columns:
    total_passed_students += df[df[subject] >= passing_marks].shape[0]
    total_failed_students += df[df[subject] < passing_marks].shape[0]

print("Overall Pass/Fail Count:")
print("="*60)
print(f"Total Students Passed : {total_passed_students} students")
print(f"Total Students Failed : {total_failed_students} students")
print("="*60)
```
