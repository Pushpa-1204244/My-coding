```python
customer_name = input("Enter the name:")
age_str = input("Enter age:")
day = input("Enter day:")


age = int(age_str)

base_price = 0
category = ""

if age < 12:
  base_price = 150
  category = "Child"
elif 12 <= age <= 59:
  base_price = 250
  category = "Adult"
elif age >= 60:
  base_price = 180
  category = "Senior Citizen"


discount_day = False
discount_amount = 0
final_price = base_price

if day.lower() == 'tuesday' or day.lower() == 'wednesday':
  discount_day = True
  discount_amount = base_price * 0.20
  final_price = base_price - discount_amount


print("Ticket Booking Summary")
print("-"*60)
print(f"Customer: {customer_name}")
print(f"Category: {category}")
print(f"Base Price: Rs. {base_price:.1f}")

if discount_day:
  print(f"Discount Day: Yes (20% off)")
  print(f"Discount Amount: Rs. {discount_amount:.1f}")
else:
  print(f"Discount Day: No")

print(f"Final Price: Rs. {final_price:.1f}")
```

```python
print("ATM WITHDRAWAL")
print("-"*60)
account_balance_str = input("Enter your Current Account Balance:")
withdrawal_amount_str = input("Enter the amount you want to withdraw:")

# Initialize variables
service_charge = 0
transaction_successful = True
error_messages = []

account_balance = 0.0 # Default value
withdrawal_amount = 0.0 # Default value
total_deducted = 0.0 # Default value

# Convert inputs to float with error handling
try:
  account_balance = float(account_balance_str)
except ValueError:
  error_messages.append("Invalid input for account balance. Please enter a number.")
  transaction_successful = False

try:
  withdrawal_amount = float(withdrawal_amount_str)
except ValueError:
  error_messages.append("Invalid input for withdrawal amount. Please enter a number.")
  transaction_successful = False

# Only proceed with further validations if conversions were successful
if transaction_successful:
  # 3. Validates all conditions
  # Rule: Minimum withdrawal: Rs. 100
  if withdrawal_amount < 100:
    error_messages.append("Withdrawal amount must be at least Rs. 100.")
    transaction_successful = False

  # Rule: Maximum withdrawal per transaction: Rs. 20,000
  if withdrawal_amount > 20000:
    error_messages.append("Maximum withdrawal per transaction is Rs. 20,000.")
    transaction_successful = False

  # Rule: Withdrawal amount must be in multiples of 100
  if withdrawal_amount % 100 != 0:
    error_messages.append("Withdrawal amount must be in multiples of 100.")
    transaction_successful = False

  # Calculate service charge and total deducted only if withdrawal_amount is valid
  if transaction_successful:
    if withdrawal_amount < 500:
      service_charge = 20

    total_deducted = withdrawal_amount + service_charge

    # Rule: Account must have sufficient balance
    if account_balance < total_deducted:
      error_messages.append("Insufficient balance.")
      transaction_successful = False

# Display the transaction result
print("\nATM Withdrawal")
if transaction_successful and not error_messages: # Ensure no errors were collected
  remaining_balance = account_balance - total_deducted

  print(f"Withdrawal amount: Rs. {withdrawal_amount:.1f}")
  if service_charge > 0:
    print(f"Service charge: Rs. {service_charge:.1f}")
  print(f"Total deducted: Rs. {total_deducted:.1f}")
  print("Transaction successful!")
  print(f"Remaining balance: Rs. {remaining_balance:.1f}")
else:
  # 4. Displays appropriate error messages if validation fails
  print("Transaction failed!")
  for msg in error_messages:
    print(f"Error: {msg}")
```
